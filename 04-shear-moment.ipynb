```python
import numpy as np
import matplotlib.pyplot as plt

# 让 matplotlib 支持中文(Windows 自带字体)
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'SimSun']
plt.rcParams['axes.unicode_minus'] = False   # 解决负号显示成方块的问题
```

```python
L = 4.0                  # 跨长,m
P = 10.0                 # 集中力,kN
x = np.linspace(0, L, 200)   # 在 0~L 之间均匀取 200 个点
```

```python
RA = RB = P / 2                                  # 支反力

V = np.where(x <= L/2, RA, -RB)                  # 剪力:左半 +5,右半 -5
M = np.where(x <= L/2, RA*x, RB*(L-x))           # 弯矩:左半 RA·x,右半 RB·(L-x)
```

```python
plt.figure(figsize=(6, 3))
plt.plot(x, V, color='blue')
plt.axhline(0, color='black', linewidth=0.8)   # 画一条水平零线
plt.xlabel('x (m)')
plt.ylabel('V (kN)')
plt.title('剪力图')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.plot(x, M, color='red')
plt.axhline(0, color='black', linewidth=0.8)
plt.xlabel('x (m)')
plt.ylabel('M (kN·m)')
plt.title('弯矩图')
plt.grid(True)
plt.show()
```
