```python
import sympy as sp

M, y, I, b, h, d = sp.symbols('M y I b h d')   # 弯矩、离中性轴距离、惯性矩、宽、高、直径
sigma = sp.symbols('sigma')
```

```python
I_rect = b * h**3 / 12
sp.Eq(I, I_rect)
```

```python
I_circ = sp.pi * d**4 / 64
sp.Eq(I, I_circ)
```

```python
sigma_expr = M * y / I
sp.Eq(sigma, sigma_expr)
```

```python
sigma_max = M * (h / 2) / I_rect
sp.Eq(sigma, sp.simplify(sigma_max))   # 应化简成 6M/(bh²)
```

```python
vals = {M: 5e6, b: 50, h: 100}
print("σ_max =", sp.N(sigma_max.subs(vals)), "MPa")
```
