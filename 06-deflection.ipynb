```python
import sympy as sp

P, L, E, I, x = sp.symbols('P L E I x')
C1, C2 = sp.symbols('C1 C2')
```

```python
M = P/2 * x                        # 左半段弯矩(跨中集中力,0≤x≤L/2)
w_dd = -M / (E * I)                # 由 EI·w'' = -M,得曲率 w''

w_d = sp.integrate(w_dd, x) + C1   # 积分一次 → 转角(斜率)
w   = sp.integrate(w_d, x) + C2    # 积分两次 → 挠度
w                                  # 显示通解(含 C1、C2)
```

```python
# 边界条件:左端 w(0)=0;跨中斜率 w'(L/2)=0(对称性)
sol = sp.solve([w.subs(x, 0), w_d.subs(x, L/2)], [C1, C2])
w_final = sp.simplify(w.subs(sol))
w_final
```

```python
w_max = sp.simplify(w_final.subs(x, L/2))
w_max        # 应化简成 P*L**3/(48*E*I)
```

```python
vals = {P: 10e3, L: 4e3, E: 2.1e5, I: 8e6}   # N、mm、MPa、mm⁴
print("最大挠度 =", sp.N(w_max.subs(vals)), "mm")
```
