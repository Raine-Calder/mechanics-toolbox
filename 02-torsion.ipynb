```python
import sympy as sp

T, G, L, d, J = sp.symbols('T G L d J')   # 扭矩、剪切模量、杆长、直径、极惯性矩
tau, phi = sp.symbols('tau phi')          # 切应力、扭转角
```

```python
J_solid = sp.pi * d**4 / 32
sp.Eq(J, J_solid)
```

```python
tau_max = T * (d / 2) / J_solid
sp.Eq(tau, sp.simplify(tau_max))
```

```python
phi_expr = T * L / (G * J_solid)
sp.Eq(phi, phi_expr)
```

```python
vals = {T: 500000, L: 500, G: 8e4, d: 40}   # T 单位 N·mm
print("τ_max =", tau_max.subs(vals), "MPa")
print("φ =", phi_expr.subs(vals), "rad")
```
