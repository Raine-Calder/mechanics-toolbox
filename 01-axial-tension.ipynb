```python
import sympy as sp

F, A, L, E = sp.symbols('F A L E')                            # 轴力、面积、杆长、弹性模量
sigma, epsilon, Delta_L = sp.symbols('sigma epsilon Delta_L') # 正应力、线应变、轴向变形
```

```python
sp.Eq(sigma, F / A)
```

```python
sp.Eq(epsilon, Delta_L / L)
```

```python
sp.Eq(sigma, E * epsilon)
```

```python
Delta_L_expr = F * L / (E * A)
sp.Eq(Delta_L, Delta_L_expr)
```

```python
vals = {F: 1000, L: 500, A: 200, E: 2.1e5}
print("σ =", (F/A).subs(vals), "MPa")  # 正应力,单位 N/mm² = MPa
print("ΔL =", Delta_L_expr.subs(vals), "mm")  # 轴向变形,单位 mm
```

```python
sol = sp.solve(sp.Eq(F/A, 5), A)      # 得到 [F/5],即 A = F/5
print("最小面积 A =", sol[0].subs({F: 1000}), "mm²")   # 200
```
