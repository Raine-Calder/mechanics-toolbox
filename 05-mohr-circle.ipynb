```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'SimSun']
plt.rcParams['axes.unicode_minus'] = False
```

```python
sigma_x = 80    # 正应力 σx,MPa
sigma_y = 20    # 正应力 σy,MPa
tau_xy  = 40    # 剪应力 τxy,MPa
```

```python
sigma_avg = (sigma_x + sigma_y) / 2                          # 圆心 σ 坐标
R = np.sqrt(((sigma_x - sigma_y)/2)**2 + tau_xy**2)          # 半径

sigma_1 = sigma_avg + R     # 最大主应力
sigma_2 = sigma_avg - R     # 最小主应力
tau_max = R                 # 最大剪应力

print("圆心 σ_avg =", sigma_avg, "MPa")
print("半径 R =", R, "MPa")
print("主应力 σ1 =", sigma_1, "MPa")
print("主应力 σ2 =", sigma_2, "MPa")
print("最大剪应力 τmax =", tau_max, "MPa")
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

# 画圆:圆心 (sigma_avg, 0),半径 R
circle = plt.Circle((sigma_avg, 0), R, fill=False, color='blue', linewidth=2)
ax.add_patch(circle)

# 两个已知应力状态点(直径两端):(σx, τxy) 和 (σy, -τxy)
ax.plot([sigma_x, sigma_y], [tau_xy, -tau_xy], 'ro', label='已知状态')

# 主应力点(在 σ 轴上)
ax.plot([sigma_1, sigma_2], [0, 0], 'go', markersize=8, label='主应力')

# 圆心
ax.plot(sigma_avg, 0, 'kx', label='圆心')

ax.set_xlabel('正应力 σ (MPa)')
ax.set_ylabel('剪应力 τ (MPa)')
ax.set_title('Mohr 圆')
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.grid(True)
ax.set_aspect('equal')   # 关键!让横纵比例 1:1,否则圆变椭圆
ax.legend()
plt.show()
```
